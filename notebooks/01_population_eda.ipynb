# 01 - Explore the synthetic population

Wallet-level EDA. Run `make data && make pipeline` first.

```python
import sys; sys.path.insert(0, '..')
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from common import settings
pd.set_option('display.width', 160)

users = pd.read_parquet(settings.PROCESSED_DIR / 'users.parquet')
panel = pd.read_parquet(settings.PROCESSED_DIR / 'monthly_panel.parquet')
feats = pd.read_parquet(settings.PROCESSED_DIR / 'financial_features.parquet')
train = pd.read_parquet(settings.PROCESSED_DIR / 'training_dataset.parquet')
print('users', len(users), '| panel', len(panel), '| features', feats.shape, '| train', train.shape)
```

```python
feats.groupby('persona')[['monthly_income','monthly_expense','savings_rate',
      'emergency_fund_months','dna_financial_health','dna_financial_risk']].median().round(2)
```

```python
feats['dna_financial_health'].hist(bins=40, figsize=(8,3))
plt.title('Financial health score'); plt.xlabel('health (0-100)'); plt.show()
```

```python
m = panel.groupby('year_month')[['income','expense','emi','net_cash_flow']].mean()
m.plot(figsize=(10,4), title='Population average monthly flows'); plt.show()
```

```python
cat_cols = [c for c in panel.columns if c.startswith('exp_')]
mix = panel[cat_cols].mean(); mix = (mix / mix.sum()).sort_values(ascending=False)
mix.plot(kind='bar', figsize=(9,3), title='Average spending mix'); plt.show()
```

```python
dims = [c for c in feats.columns if c.startswith('dna_')]
feats.groupby('segment_label')[dims].mean().round(1)
```

```python
print('stress rate:', round(train['stress_label'].mean(), 3))
train.groupby('persona')['stress_label'].mean().sort_values(ascending=False).round(3)
```
